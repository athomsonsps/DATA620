For this assignment, I created a basic network graph based on the Krackhardt kite example. I used Python and NetworkX to create the nodes and edges, then used Matplotlib to display the graph.


## 1. Import the packages

NetworkX is used to build the graph and Matplotlib is used to display it.


```python
import networkx as nx
import matplotlib.pyplot as plt
```

## 2. Create the nodes

Each person in the network is represented by a node.


```python
nodes = [
    "Andre", "Beverly", "Carol", "Diane", "Ed",
    "Fernando", "Garth", "Heather", "Ike", "Jane"
]

G = nx.Graph()
G.add_nodes_from(nodes)

print("Number of nodes:", G.number_of_nodes())
print("Nodes:", list(G.nodes()))
```

## 3. Create the edges

The edges represent the connections between the people in the network.


```python
edges = [
    ("Andre", "Beverly"),
    ("Andre", "Carol"),
    ("Andre", "Diane"),
    ("Andre", "Fernando"),
    ("Beverly", "Diane"),
    ("Beverly", "Ed"),
    ("Beverly", "Garth"),
    ("Carol", "Diane"),
    ("Carol", "Fernando"),
    ("Diane", "Ed"),
    ("Diane", "Fernando"),
    ("Diane", "Garth"),
    ("Ed", "Garth"),
    ("Fernando", "Garth"),
    ("Fernando", "Heather"),
    ("Garth", "Heather"),
    ("Heather", "Ike"),
    ("Ike", "Jane")
]

G.add_edges_from(edges)

print("Number of edges:", G.number_of_edges())
print("Edges:", list(G.edges()))
```

## 4. Draw the graph

I used fixed positions to keep the graph close to the kite shape shown in the assignment.


```python
pos = {
    "Andre": (0, 2),
    "Beverly": (0, 0),
    "Carol": (1, 3),
    "Diane": (1, 1.5),
    "Ed": (1, -1),
    "Fernando": (2, 2),
    "Garth": (2, 0),
    "Heather": (3, 1),
    "Ike": (4, 1),
    "Jane": (5, 1)
}

plt.figure(figsize=(10, 6))
nx.draw(
    G,
    pos,
    with_labels=True,
    node_size=1800,
    font_size=10,
    font_weight="bold",
    width=1.5
)
plt.title("Krackhardt Kite Graph")
plt.axis("off")
plt.show()
```

## 5. Check the graph

The graph has 10 nodes and 18 edges. Most of the connections are on the left side of the graph. Heather connects that group to the chain leading to Ike and Jane.


```python
print(f"The graph has {G.number_of_nodes()} nodes and {G.number_of_edges()} edges.")
print("Connected:", nx.is_connected(G))
```
